```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# seaborn has built-in datasets :
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset('penguins')
```

```python
#  scatter plot :
# hue allows the to differentiate data on the basis of their catecories. like here, it differentiates on the basis of island, sex will differentiate on the basis of gender : male/female.
# set_style():
sns.set_style("darkgrid")
sns.set_context("talk")
sns.scatterplot(data= penguins, x = "flipper_length_mm", y= "body_mass_g", hue="sex", palette="Set3")
```

```python

sns.set_style("dark")
sns.set_context("notebook") # default
sns.scatterplot(data= penguins, x = "flipper_length_mm", y= "body_mass_g", hue="island", palette="Set1")
```

```python

sns.set_style("darkgrid")
sns.set_context("paper")
sns.scatterplot(data= penguins, x = "flipper_length_mm", y= "body_mass_g", hue="island", palette="Dark2") 
```

```python
# The different ways for styling are as follows :
# white, dark, whitegrid, darkgrid, ticks

# set_context() -> allows us to override default parameters. this affects things like the size of the labels, lines and other elements of your plot, but not the overall style -> set_style(). The context() are:
# 1. poster
# 2. paper -> good for printing academic papers.
# 3. Notebook -> good for working inside jupyter notebook (default setting).
# 4. Talk -> good for presentation slides where readability matters.
```

```python
sns.set_style("whitegrid")
sns.set_context("paper")
sns.scatterplot(data= penguins, x="species", y="body_mass_g", hue="island", style="sex", alpha=0.5)
# here style=sex means the data is being grouped according to gender, male = dot, female = x.
# alpha will decrease the opacity of the data points to aoid oerlapping.
```

```python
# stripplot :

sns.stripplot(data=penguins, x="species", y="body_mass_g", hue="island", dodge=True, alpha=0.5, jitter=True)
# dodge = True will sort the datapoints on the basis of hue.
# jitter = True will allow us to spread the datapoints more.
```

```python
# swarmplot :
sns.swarmplot(data=penguins, x="species", y="body_mass_g", hue="island", alpha=0.7, palette="Dark2")
```

```python
# histogram :
sns.histplot(data= penguins, x="body_mass_g", hue="sex", multiple="stack") #multiple is used to display the histogram properly.
```

```python
# regression plot :
sns.set_style("whitegrid")
sns.regplot(data=penguins, x="body_mass_g", y="flipper_length_mm", color="purple")
```

```python
# line plot:
sns.lineplot(data=penguins, x="body_mass_g", y="flipper_length_mm", hue="island", style="sex", palette=["purple","blue","pink"])
```

```python
# jointplot : A jointplot in Seaborn is a plot that lets you see the relationship between two variables AND the distribution of each variable at the same time.Think of it as: Scatter plot in the middle + distribution plots on the sides.

sns.jointplot(data=penguins, x="body_mass_g", y="flipper_length_mm", hue="sex", palette=["purple","pink"])
```

```python
sns.jointplot(data=penguins, x="body_mass_g", y="flipper_length_mm", hue="sex", palette=["blue","red"], kind="kde")
```

```python
sns.barplot(data=penguins, x="species", y="body_mass_g", hue="sex", palette=["red","pink"], estimator=np.mean)
#  mean distribution of species. (calculated avg bodymass)
#  if we do estimator=np.sum the scale is changed. (np.mean is already applied by default).
```

```python
# countplot : direct implementation to count something without using alue_counts()

sns.countplot(data=penguins, x="species", hue="island", palette={"lightblue","pink","lavender"})
```

<!-- View box plot with explanation: ![alt text](image.png) -->

```python
# box plot: used to plot extreme high and extreme low values
sns.boxplot(data=penguins, x="species", y="body_mass_g", hue="sex", palette=["red","pink"])
```

<!-- View violin plot with explanation:  ![alt text](image-1.png) -->

```python
# violin plot: it used kernal density estimation.
# lower adj value = minimum alue, upper adj value = maximum value, middle line = median, part bellow the median = Q1, part above the median = Q3, part above the upper adj value = outliers/outside points, the line above Q3 = whiskers(Range), thick bar = IQR (if inner 'box') Q3-Q1, width of the violin = density of the data at different values.

sns.violinplot(data=penguins, x="species", y="body_mass_g", hue="sex", palette=["blue","pink"], split=True, inner='box')
```

```python
# KDE plot: kernal density estimation plot used for estimating probability density function of continuous or non parametric data. (A smooth curve that shows the distribution of data like a smoothed out version). it helps you to see where the data is concentrated (peaks) and how it spreads.

sns.kdeplot(data=penguins, x="body_mass_g", hue="species", fill=True)
```

```python
# heatmap: 2D color-coded matrix that shows the relationship or patterns between two or more varibleds.
"""
It is commonly used to:
1. Show correlation between features.
2. Visualize confusion matrices.
3. Highlight missing data -> sns.heatmap(df.isnull()).

"""
penguins.head()
```

```python
cols = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]

# calculate the correlation between these cols :

penguins[cols].corr()
```

```python
sns.heatmap(data=penguins[cols].corr(), annot=True, cmap="Blues", vmin=-0.2)
```

```python
# Rug plot:
sns.rugplot(data=penguins, x="body_mass_g", hue="species",palette="Set2", height=0.2)
```

```python
sns.rugplot(data=penguins, y="body_mass_g", hue="species",palette="Set2", height=0.2)
```

```python
#  pair plot: grid of plots that shos the relationship between all the numerical variables in a data frame. it automatically filters out the categorical columns and only works with the numerical columns in a df. so even if we pass the entire df at once it wont raise an error like a heatmap.

sns.pairplot(data=penguins, hue="sex", palette="Set2", diag_kind="kde")
```

```python
# pairgrid: subplot grid for plotting pairwise relationships in a dataset.
graph = sns.PairGrid(data=penguins, hue="sex", palette="Set2")
graph.map_upper(sns.scatterplot) #upper triangle
graph.map_lower(sns.kdeplot) #lower triangle
graph.map_diag(sns.histplot) #diagonal
graph.add_legend()
```
