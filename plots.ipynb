```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# seaborn has built-in datasets :
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.head()
```

```python
print(penguins.groupby("island")["species"].value_counts())
```

```python
penguins['species'].value_counts()
```

```python
penguins['island'].value_counts()
```

```python
#  scatter plot :
# hue allows the to differentiate data on the basis of their catecories. like here, it differentiates on the basis of island, sex will differentiate on the basis of gender : male/female.
# set_style():
sns.set_style("darkgrid")
sns.set_context("talk")
sns.scatterplot(data= penguins, x = "flipper_length_mm", y= "body_mass_g", hue="sex", palette="Set3")
```

```python

sns.set_style("dark")
sns.set_context("notebook") # default
sns.scatterplot(data= penguins, x = "flipper_length_mm", y= "body_mass_g", hue="island", palette="Set1")
```

```python

sns.set_style("darkgrid")
sns.set_context("paper")
sns.scatterplot(data= penguins, x = "flipper_length_mm", y= "body_mass_g", hue="island", palette="Dark2") 
```

```python
# The different ways for styling are as follows :
# white, dark, whitegrid, darkgrid, ticks

# set_context() -> allows us to override default parameters. this affects things like the size of the labels, lines and other elements of your plot, but not the overall style -> set_style(). The context() are:
# 1. poster
# 2. paper -> good for printing academic papers.
# 3. Notebook -> good for working inside jupyter notebook (default setting).
# 4. Talk -> good for presentation slides where readability matters.
```

```python
sns.set_style("whitegrid")
sns.set_context("paper")
sns.scatterplot(data= penguins, x="species", y="body_mass_g", hue="island", style="sex", alpha=0.5)
# here style=sex means the data is being grouped according to gender, male = dot, female = x.
# alpha will decrease the opacity of the data points to aoid oerlapping.
```

```python
# stripplot :

sns.stripplot(data=penguins, x="species", y="body_mass_g", hue="island", dodge=True, alpha=0.5, jitter=True)
# dodge = True will sort the datapoints on the basis of hue.
# jitter = True will allow us to spread the datapoints more.
```

```python
# swarmplot :
sns.swarmplot(data=penguins, x="species", y="body_mass_g", hue="island", alpha=0.7, palette="Dark2")
```

```python
# histogram :
sns.histplot(data= penguins, x="body_mass_g", hue="sex", multiple="stack") #multiple is used to display the histogram properly.
```

```python
# regression plot :
sns.set_style("whitegrid")
sns.regplot(data=penguins, x="body_mass_g", y="flipper_length_mm", color="purple")
```

```python
# line plot:
sns.lineplot(data=penguins, x="body_mass_g", y="flipper_length_mm", hue="island", style="sex", palette=["purple","blue","pink"])
```

```python
# jointplot : A jointplot in Seaborn is a plot that lets you see the relationship between two variables AND the distribution of each variable at the same time.Think of it as: Scatter plot in the middle + distribution plots on the sides.

sns.jointplot(data=penguins, x="body_mass_g", y="flipper_length_mm", hue="sex", palette=["purple","pink"])
```

```python
sns.jointplot(data=penguins, x="body_mass_g", y="flipper_length_mm", hue="sex", palette=["blue","red"], kind="kde")
```

```python
sns.barplot(data=penguins, x="species", y="body_mass_g", hue="sex", palette=["red","pink"], estimator=np.mean)
#  mean distribution of species. (calculated avg bodymass)
#  if we do estimator=np.sum the scale is changed. (np.mean is already applied by default).
```

```python
# countplot : direct implementation to count something without using alue_counts()

sns.countplot(data=penguins, x="species", hue="island", palette={"lightblue","pink","lavender"})
```
